```python
# 定义工具

from langchain.tools import tool

@tool  # 装饰器
def get_weather(location: str) -> str:
    """
    获取某个地方的天气
    :param location:
    :return:
    """
    return f"current weather in {location} is rainy"
```

```python
# 创建 Agent

from langchain.agents import create_agent

agent = create_agent(
    "deepseek-v4-pro",
    tools = [get_weather]  # 工具，数组形式，可以有多个工具
)
```

```python
# 发起调用

print("正在调用deepseek大模型")

response = agent.invoke(
    {
        "messages":[
            {"role": "user", "content": "今天北京的天气如何？"}
        ]
    }
)
```
