```python
# 导入LangChain的初始化模型的函数

from langchain.chat_models import init_chat_model
```

```python
# 调用 init_chat_model 函数初始化模型
model = init_chat_model(model = "deepseek-v4-flash")
```

```python
print(type(model))
```
